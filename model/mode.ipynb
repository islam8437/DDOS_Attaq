```python
from scipy.stats import f_oneway
import pandas as pd

# Vérifier si Y et df_filtered sont bien alignés
Y = pd.Series(Y, index=df_filtered.index)  # Assurer l'alignement

# Stocker les résultats dans un dictionnaire
results = {}

for col in df_filtered.columns:
    # Créer des groupes selon les valeurs uniques de Y
    groupes = [df_filtered.loc[Y == val, col].values for val in Y.unique()]
    
    # Vérifier s'il y a assez d'échantillons dans chaque groupe
    if all(len(g) > 1 for g in groupes):  
        stat, p_value = f_oneway(*groupes)
        results[col] = (stat, p_value)
        print(f"Feature {col} - F-statistique: {stat:.3f}, p-value: {p_value:.5f}")
    else:
        print(f"Feature {col} ignorée (pas assez de valeurs uniques dans Y)")

# Afficher les résultats sous forme de DataFrame
print(results)
```
